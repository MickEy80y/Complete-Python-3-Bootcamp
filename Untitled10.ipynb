```python

# This Python 3 environment comes with many helpful analytics libraries installed
# It is defined by the kaggle/python Docker image: https://github.com/kaggle/docker-python
# For example, here's several helpful packages to load

import numpy as np # linear algebra
import pandas as pd # data processing, CSV file I/O (e.g. pd.read_csv)

# Input data files are available in the read-only "../input/" directory
# For example, running this (by clicking run or pressing Shift+Enter) will list all files under the input directory

import os
for dirname, _, filenames in os.walk('/kaggle/input'):
    for filename in filenames:
        print(os.path.join(dirname, filename))

# You can write up to 20GB to the current directory (/kaggle/working/) that gets preserved as output when you create a version using "Save & Run All"
# You can also write temporary files to /kaggle/temp/, but they won't be saved outside of the current session
import matplotlib.pyplot as plt
import numpy as np

# Define the size of the checkerboard
n = 10

# Create a checkerboard pattern using NumPy array manipulation
checkerboard = np.zeros((n, n), dtype=int)
checkerboard[1::2, ::2] = 1
checkerboard[::2, 1::2] = 1

# Display the checkerboard using imshow with a grayscale colormap
plt.imshow(checkerboard, cmap='ocean')
plt.title('Checkerboard Pattern')
plt.xticks([]) # Remove x-axis ticks
plt.yticks([]) # Remove y-axis ticks
plt.grid(False) # Ensure no grid lines are drawn
plt.show()
#That a very organized and user-friendly approach! Creating a single function that can calculate either the domain or the range based on a user's choice makes the code more modular and easier to use.
#Hee's how you could structure such a function using if-else statements to select between domain and range calculation:
from sympy import symbols, sqrt, log, Abs
from sympy.calculus.util import continuous_domain
from sympy.sets import Reals
from sympy import diff, solve, oo

x = symbols('x')

def evaluate_function_safe(function, value):
    """Evaluates a symbolic function safely."""
    try:
        result = function.subs(x, value)
        return result
    except ZeroDivisionError:
        return float('nan')
    except Exception:
        return None

def find_domain(expression, variable):
    """Finds the domain of a symbolic expression."""
    return continuous_domain(expression, variable, Reals)

def find_range_calculus(function, variable, interval=None):
    """Attempts to find the range using calculus."""
    derivative = diff(function, variable)
    critical_points = solve(derivative, variable)
    potential_extrema_values = []
    for cp in critical_points:
        value = evaluate_function_safe(function, cp)
        if value is not None:
            potential_extrema_values.append(value)
    if interval:
        a, b = interval
        val_a = evaluate_function_safe(function, a)
        val_b = evaluate_function_safe(function, b)
        if val_a is not None:
            potential_extrema_values.append(val_a)
        if val_b is not None:
            potential_extrema_values.append(val_b)
        valid_values = [val for val in potential_extrema_values if not val is None and not val is float('nan')]
        if valid_values:
            min_val = min(valid_values)
            max_val = max(valid_values)
            return f"Range on [{a}, {b}]: [{min_val}, {max_val}]"
        else:
            return f"Could not determine range on [{a}, {b}] reliably."
    else:
        limit_pos_inf = function.limit(variable, oo)
        limit_neg_inf = function.limit(variable, -oo)
        return f"Potential values: {potential_extrema_values}, Limits at +/- infinity: ({limit_neg_inf}, {limit_pos_inf})"

def analyze_function(expression, variable, task="domain", interval=None):
    """
    Analyzes a function to find either its domain or range.

    Args:
        expression: The symbolic expression of the function.
        variable: The symbolic variable.
        task: "domain" to find the domain, "range" to attempt to find the range.
        interval: Optional interval (tuple) to find the range over.

    Returns:
        The domain or a description of the range attempt.
    """
    if task.lower() == "domain":
        return find_domain(expression, variable)
    elif task.lower() == "range":
        return find_range_calculus(expression, variable, interval)
    else:
        return "Invalid task. Choose 'domain' or 'range'."

# Example usage:
f = x**2 / (x - 1)

# Find the domain
domain_f = analyze_function(f, x, task="domain")
print(f"\nDomain of {f}: {domain_f}")

# Attempt to find the range over an interval
range_f_interval = analyze_function(f, x, task="range", interval=(-2, 0))
print(f"Range of {f} on [-2, 0]: {range_f_interval}")

# Attempt to find the general range
range_f_general = analyze_function(f, x, task="range")
print(f"General range analysis of {f}: {range_f_general}")

g = sqrt(4 - x**2)

# Find the domain of g
domain_g = analyze_function(g, x, task="domain")
print(f"\nDomain of {g}: {domain_g}")

# Attempt to find the range of g over its domain (approximately)
# For simplicity here, we'll just try a reasonable interval within the domain
range_g_interval = analyze_function(g, x, task="range", interval=(-2, 2))
print(f"Range of {g} on [-2, 2]: {range_g_interval}")

##In this analyze_function, the task argument determines whether to call find_domain or find_range_calculus. This keeps the functionality grouped logically.
#What do you think of this structure? Would you like to add any other features or modify it?
```
